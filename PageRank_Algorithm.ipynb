```python
import numpy as np
```

```python
class PageRank:
  def __init__(self, damping_factor = 0.85, max_iter = 1000):
    self.df = damping_factor
    self.iterations = max_iter




  def build_matrix(self, edges, nodes):

    index = {node : i for i, node in enumerate(nodes)}
    n = len(nodes)

    matrix  = np.zeros((n, n))

    for u, v in edges:
      matrix[index[u], index[v]] = 1

    return matrix, index




  def normalize(self, matrix):

    r_total = matrix.sum(axis = 1)

    for i in range(len(r_total)):
      if r_total[i] != 0:
        matrix[i,:] /= r_total[i]

    return matrix




  def compute(self, edges, nodes):

    matrix, index = self.build_matrix(edges, nodes)
    matrix = self.normalize(matrix)

    n = len(nodes)

    p_rank = np.ones(n)/n
    iter = 0
    p_rank_init = np.zeros(n)
    while iter < self.iterations and np.max(np.abs(p_rank - p_rank_init)) > 1e-15:
      p_rank_init = p_rank
      p_rank = ((1 - self.df)/n) + (self.df * np.dot(matrix.T, p_rank))
      iter += 1

    results = {}

    for node, i in index.items():
      results[node] = p_rank[i]

    return results, iter
```

```python
nodes = ['A', 'B', 'C', 'D']

edges = [
    ('A', 'B'),
    ('A', 'C'),
    ('B', 'C'),
    ('C', 'A'),
    ('D', 'C')
]
pr = PageRank()
results, iter = pr.compute(edges, nodes)
iter
```
